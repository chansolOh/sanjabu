```python
import os
import shutil
import numpy as np 


root_path = "/nas/Dataset/Dataset_2025/dataset_v1"
target_path = "/home/cubox/ochansol/isaac_code/python/sanjabu/2025/evaluation/Data_samples"
sampling_num = 100  

env_list = [i for i in os.listdir(root_path) if os.path.isdir(os.path.join(root_path, i))]
for env in env_list:
    section_list = [i for i in os.listdir(os.path.join(root_path, env)) if os.path.isdir(os.path.join(root_path, env, i))]
    for section in section_list:
        platform_list = [i for i in os.listdir(os.path.join(root_path, env, section)) if os.path.isdir(os.path.join(root_path, env, section, i))]
        for platform in platform_list:
            print(f"Processing Environment: {env} | Section: {section} | Platform: {platform}")
            rgb_list = [i for i in os.listdir(os.path.join(root_path, env, section, platform, "rgb", "top_view_camera")) if i.endswith(".png")]
            rgb_sample_list = np.random.choice(rgb_list, sampling_num, replace=False)
            for rgb in rgb_sample_list:
                src_rgb_path            = os.path.join(root_path, env, section, platform, "rgb", "top_view_camera",         rgb)
                src_depth_path          = os.path.join(root_path, env, section, platform, "depth", "top_view_camera",       rgb.replace(".png", ".npy"))
                src_inst_seg_path       = os.path.join(root_path, env, section, platform, "inst_seg", "top_view_camera",    rgb)
                src_inst_seg_json_path  = os.path.join(root_path, env, section, platform, "inst_seg", "top_view_camera",    f"semantics_mapping_{rgb.replace('.png', '')}.json")
                src_bbox_path           = os.path.join(root_path, env, section, platform, "bbox", "top_view_camera",        rgb.replace(".png", ".json"))
                src_normals_path        = os.path.join(root_path, env, section, platform, "normals", "top_view_camera",     rgb)
                src_scene_meta_path     = os.path.join(root_path, env, section, platform, "scene_meta",                     rgb.replace(".png", ".json"))
                src_grasp_path          = os.path.join(root_path, env, section, platform, "output_grasp",                   rgb.replace(".png", ".json"))

                dst_rgb_dir_path            = os.path.join(target_path, env, section, platform, "rgb", "top_view_camera")
                dst_depth_dir_path          = os.path.join(target_path, env, section, platform, "depth", "top_view_camera")
                dst_inst_seg_dir_path       = os.path.join(target_path, env, section, platform, "inst_seg", "top_view_camera")
                dst_inst_seg_json_dir_path  = os.path.join(target_path, env, section, platform, "inst_seg", "top_view_camera")
                dst_bbox_dir_path           = os.path.join(target_path, env, section, platform, "bbox", "top_view_camera")
                dst_normals_dir_path        = os.path.join(target_path, env, section, platform, "normals", "top_view_camera")
                dst_scene_meta_dir_path     = os.path.join(target_path, env, section, platform, "scene_meta")
                dst_grasp_dir_path          = os.path.join(target_path, env, section, platform, "output_grasp")

                dst_rgb_path            = os.path.join(target_path, env, section, platform, "rgb", "top_view_camera",         rgb)
                dst_depth_path          = os.path.join(target_path, env, section, platform, "depth", "top_view_camera",       rgb.replace(".png", ".npy"))
                dst_inst_seg_path       = os.path.join(target_path, env, section, platform, "inst_seg", "top_view_camera",    rgb)
                dst_inst_seg_json_path  = os.path.join(target_path, env, section, platform, "inst_seg", "top_view_camera",    f"semantics_mapping_{rgb.replace('.png', '')}.json")
                dst_bbox_path           = os.path.join(target_path, env, section, platform, "bbox", "top_view_camera",        rgb.replace(".png", ".json"))
                dst_normals_path        = os.path.join(target_path, env, section, platform, "normals", "top_view_camera",     rgb)
                dst_scene_meta_path     = os.path.join(target_path, env, section, platform, "scene_meta",                     rgb.replace(".png", ".json"))
                dst_grasp_path          = os.path.join(target_path, env, section, platform, "output_grasp",                   rgb.replace(".png", ".json"))
                
                os.makedirs(dst_rgb_dir_path, exist_ok=True)
                os.makedirs(dst_depth_dir_path, exist_ok=True)
                os.makedirs(dst_inst_seg_dir_path, exist_ok=True)
                os.makedirs(dst_bbox_dir_path, exist_ok=True)
                os.makedirs(dst_normals_dir_path, exist_ok=True)
                os.makedirs(dst_scene_meta_dir_path, exist_ok=True)
                os.makedirs(dst_grasp_dir_path, exist_ok=True)
                
                
                shutil.copyfile(src_rgb_path, dst_rgb_path)
                shutil.copyfile(src_depth_path, dst_depth_path)
                shutil.copyfile(src_inst_seg_path, dst_inst_seg_path)
                shutil.copyfile(src_inst_seg_json_path, dst_inst_seg_json_path)
                shutil.copyfile(src_bbox_path, dst_bbox_path)
                shutil.copyfile(src_normals_path, dst_normals_path)
                shutil.copyfile(src_scene_meta_path, dst_scene_meta_path)
                shutil.copyfile(src_grasp_path, dst_grasp_path)



            

```

# platform path edit

```python
import json
import os

root_path = "/nas/Dataset/Dataset_2025/dataset_v1"
env_dir_list = os.listdir(root_path)
for env in env_dir_list:
    section_dir_list = os.listdir(os.path.join(root_path, env))
    for section in section_dir_list:
        platform_dir_list = os.listdir(os.path.join(root_path, env, section))
        for platform in platform_dir_list:
            if len(platform.split(".")) ==1 and platform not in ["conf","rgb","depth", "bbox", "normals", "pointcloud", "inst_seg"]:
                conf_list = os.listdir(os.path.join(root_path, env, section, platform,"conf"))
                for conf in conf_list:
                    if conf.endswith(".json"):
                        with open(os.path.join(root_path, env, section, platform, "conf", conf), "r") as f:
                            data = json.load(f)
                        if section not in data["platform"]["usd_path"]:
                            split = data["platform"]["usd_path"].split("platform_usd")
                            data["platform"]["usd_path"] = split[0] + "platform_usd/" + section + split[1]
                            # print(data["platform"]["usd_path"])
                            with open(os.path.join(root_path, env, section, platform, "conf", conf), "w") as f:
                                json.dump(data, f, indent=4)
                print(platform," - comp")
    print(env, " - comp")
```

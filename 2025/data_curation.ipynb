```python
import os
import shutil
from tqdm import tqdm
root_path =         "/nas/Dataset/Dataset_2025/dataset_v1"
root_output_path =  "/nas/Dataset/Dataset_2025/dataset_v1_filter_out"
cam_path_list = ["top_view_camera", "side_view_camera"]

def move_file(file_name, pre_path, out_path, ext_path):
    if not os.path.exists(ext_path):
        shutil.move(os.path.join(pre_path,file_name), os.path.join(out_path,file_name))

for env_name in os.listdir(root_path):
    if env_name.endswith(".zip") or env_name.endswith(".tar"):
        continue
    for section_name in os.listdir(os.path.join(root_path, env_name)):
        for platform_name in tqdm(os.listdir(os.path.join(root_path, env_name, section_name))):
            pre_path = os.path.join(root_path, env_name, section_name, platform_name)
            out_path = os.path.join(root_output_path, env_name, section_name, platform_name)

            for cam_name in cam_path_list:
                if not os.path.exists(os.path.join(out_path,"rgb", cam_name)):
                    os.makedirs(os.path.join(out_path,"rgb", cam_name))
                for file_name in os.listdir(os.path.join(pre_path, "rgb", cam_name)):
                    num = file_name.strip(".png")
                    move_file(file_name=file_name, 
                              pre_path = os.path.join(pre_path,"rgb", cam_name), 
                              out_path = os.path.join(out_path,"rgb", cam_name),
                              ext_path=  os.path.join(pre_path,"output_grasp",f"{num}.json"))
                    
            for cam_name in cam_path_list:
                if not os.path.exists(os.path.join(out_path,"depth", cam_name)):
                    os.makedirs(os.path.join(out_path,"depth", cam_name))
                for file_name in os.listdir(os.path.join(pre_path, "depth", cam_name)):
                    num = file_name.strip(".npy")
                    move_file(file_name=file_name, 
                              pre_path = os.path.join(pre_path,"depth", cam_name), 
                              out_path = os.path.join(out_path,"depth", cam_name),
                              ext_path=  os.path.join(pre_path,"output_grasp",f"{num}.json"))
                    
            for cam_name in cam_path_list:
                if not os.path.exists(os.path.join(out_path,"bbox", cam_name)):
                    os.makedirs(os.path.join(out_path,"bbox", cam_name))
                for file_name in os.listdir(os.path.join(pre_path, "bbox", cam_name)):
                    num = file_name.strip(".json")
                    move_file(file_name=file_name, 
                              pre_path = os.path.join(pre_path,"bbox", cam_name), 
                              out_path = os.path.join(out_path,"bbox", cam_name),
                              ext_path=  os.path.join(pre_path,"output_grasp",f"{num}.json"))
                    

            if not os.path.exists(os.path.join(out_path,"conf")):
                os.makedirs(os.path.join(out_path,"conf"))
            for file_name in os.listdir(os.path.join(pre_path, "conf")):
                num = file_name.strip(".json")
                move_file(file_name=file_name, 
                            pre_path = os.path.join(pre_path,"conf"), 
                            out_path = os.path.join(out_path,"conf"),
                            ext_path=  os.path.join(pre_path,"output_grasp",f"{num}.json"))
                    
            for cam_name in cam_path_list:
                if not os.path.exists(os.path.join(out_path,"inst_seg", cam_name)):
                    os.makedirs(os.path.join(out_path,"inst_seg", cam_name))
                for file_name in os.listdir(os.path.join(pre_path, "inst_seg", cam_name)):
                    if file_name.endswith(".png"):
                        num = file_name.strip(".png")
                        move_file(file_name=file_name, 
                                pre_path = os.path.join(pre_path,"inst_seg", cam_name), 
                                out_path = os.path.join(out_path,"inst_seg", cam_name),
                                ext_path=  os.path.join(pre_path,"output_grasp",f"{num}.json"))
                    elif file_name.endswith(".json"):
                        num = file_name.strip(".json").split("_")[-1]
                        move_file(file_name=file_name, 
                                pre_path = os.path.join(pre_path,"inst_seg", cam_name), 
                                out_path = os.path.join(out_path,"inst_seg", cam_name),
                                ext_path=  os.path.join(pre_path,"output_grasp",f"{num}.json"))
            
            for cam_name in cam_path_list:
                if not os.path.exists(os.path.join(out_path,"normals", cam_name)):
                    os.makedirs(os.path.join(out_path,"normals", cam_name))
                for file_name in os.listdir(os.path.join(pre_path, "normals", cam_name)):
                    num = file_name.strip(".png")
                    move_file(file_name=file_name, 
                              pre_path = os.path.join(pre_path,"normals", cam_name), 
                              out_path = os.path.join(out_path,"normals", cam_name),
                              ext_path=  os.path.join(pre_path,"output_grasp",f"{num}.json"))

            if not os.path.exists(os.path.join(out_path,"scene_meta")):
                os.makedirs(os.path.join(out_path,"scene_meta"))
            for file_name in os.listdir(os.path.join(pre_path, "scene_meta")):
                num = file_name.strip(".json")
                move_file(file_name=file_name, 
                            pre_path = os.path.join(pre_path,"scene_meta"), 
                            out_path = os.path.join(out_path,"scene_meta"),
                            ext_path=  os.path.join(pre_path,"output_grasp",f"{num}.json"))

                
                   
```

# Insert to Data

```python
import json
import csv
import os
from tqdm import tqdm
# model_path = [  "/nas/ochansol/3d_model/peel3_scan_data_2024",
#                 "/nas/ochansol/3d_model/peel3_scan_data_2025"]
root_path = "/nas/Dataset/Dataset_2025/dataset_v1"
root_path_scene_meta = "/nas/Dataset/Dataset_2025/dataset_v1"

# with open(os.path.join(model_path[0], "2024_objects_cat_attr.csv"), "r") as f:
#     model_csv_2024 = csv.DictReader(f)
#     model_csv_2024 = list(model_csv_2024)
# with open(os.path.join(model_path[1], "2025_objects_cat_attr.csv"), "r") as f:
#     model_csv_2025 = csv.DictReader(f)
#     model_csv_2025 = list(model_csv_2025)


# model_sheet = model_csv_2024 + model_csv_2025
with open("/nas/Dataset/Dataset_2025/objects_metadata.csv", "r") as f:
    model_sheet = csv.DictReader(f)
    model_sheet = list(model_sheet)
model_dict = {}
for item in model_sheet:
    model_dict[item["Class_name"]] = item
```

```python
for env_name in os.listdir(root_path):
    env_path = os.path.join(root_path, env_name)
    if not os.path.isdir(env_path):
        continue
    for section_name in os.listdir(env_path):
        section_path = os.path.join(env_path, section_name)
        if not os.path.isdir(section_path):
            continue
        for platform_name in os.listdir(section_path):
            platform_path = os.path.join(section_path, platform_name)
            if not os.path.isdir(platform_path):
                continue

            os.makedirs(os.path.join(platform_path.replace(root_path,root_path_scene_meta), "scene_meta"),exist_ok=True)
            print(f"Processing {env_name}/ {section_name}/ {platform_name} ...")
            for file_name in tqdm(os.listdir(os.path.join(platform_path,"conf"))):
                if not file_name.endswith(".json"):
                    continue
                file_path = os.path.join(platform_path, "conf", file_name)
                with open(file_path, "r") as f:
                    data = json.load(f)
                obj_data = data["objects"]
                scene_meta_dict = {
                    "objects": {}, 
                    "Description" :""}
                for obj in obj_data:
                    each_data = model_dict[obj["class"]]
                    scene_meta_dict["objects"][obj["class"]] = {
                        "level_1" : each_data["Level_1"],
                        "level_2" : each_data["Level_2"],
                        "level_3" : each_data["Level_3"],
                        "object_name" : each_data["Object_name"],
                        "color" : each_data["Color"],
                        "packaging" : each_data["Packaging"],
                        "features" : each_data["Features"],
                        "description" : each_data["Description"],
                    }
                    scene_meta_dict["Description"] += " "+each_data["Description"]
                
                with open(os.path.join(platform_path.replace(root_path,root_path_scene_meta), "scene_meta", file_name), "w") as f:
                    json.dump(scene_meta_dict, f, indent=4, ensure_ascii=False)

```

```python
conf_path = "/nas/ochansol/3d_model/peel3_scan_data_2024/objects_conf.json"
with open(conf_path, "r") as f:
    conf_data = json.load(f)
new_conf = []
for conf in conf_data:
    each_data = model_dict[conf["name"]]
    conf["attributes"] = {
        "color" : each_data["Color"],
        "packaging" : each_data["Packaging"],
        "features" : each_data["Features"],
        "description" : each_data["Description"],
    }
    conf["category"] = {
        "level_1" : each_data["Level_1"],
        "level_2" : each_data["Level_2"],
        "level_3" : each_data["Level_3"],
        "object_name" : each_data["Object_name"],
    }
    new_conf.append(conf)
```

```python
import csv
import json
csv_path = "/nas/Dataset/Dataset_2025/objects_metadata.csv"

with open(csv_path, "r") as f:
    csv_file = csv.DictReader(f)
    csv_file = list(csv_file)

csv_object_name_dict = {}
csv_old_name_dict = {}

for item in csv_file:
    csv_object_name_dict[item["Object_name"]] = item
    csv_old_name_dict[item["Old_name"]] = item
```

```python
import json
from tqdm import tqdm
import os

data_root_path ="/nas/Dataset/Dataset_2025/dataset_v1"


# for env_name in os.listdir(data_root_path):
env_name = "Home"
env_path = os.path.join(data_root_path, env_name)

# if not os.path.isdir(env_path):
#     continue
for section_name in os.listdir(env_path):
    section_path = os.path.join(env_path, section_name)
    if not os.path.isdir(section_path):
        continue
    for platform_name in os.listdir(section_path)[5:6]:
        platform_path = os.path.join(section_path, platform_name)
        print(platform_name)
        if not os.path.isdir(platform_path):
            continue
        target_dir_path = os.path.join(platform_path,"conf")
        for file_name in tqdm(sorted(os.listdir(target_dir_path))):
            if file_name.endswith(".json"):
                with open(os.path.join(target_dir_path, file_name), 'r') as f:
                    data = json.load(f)

                for idx, obj in enumerate(data["objects"]):
                    old_name = data["objects"][idx]["class"]
                    if old_name in csv_old_name_dict:
                        new_name = csv_old_name_dict[old_name]["Class_name"]
                    elif old_name in csv_object_name_dict:
                        new_name = csv_object_name_dict[old_name]["Class_name"]
                    else:
                        continue
                    data["objects"][idx]["class"] = new_name
                    data["objects"][idx]["usd_path"] = obj["usd_path"].replace(old_name, new_name)
                with open(os.path.join(target_dir_path, file_name), 'w') as f:
                    json.dump(data, f, indent=4)


                ##### bbox top
                target_dir_path_sub = os.path.join(platform_path,"bbox","top_view_camera")
                with open(os.path.join(target_dir_path_sub, file_name), 'r') as f:
                    data = json.load(f)
                keys = list(data.keys())
                for key in keys:
                    if key in csv_old_name_dict:
                        new_name = csv_old_name_dict[key]["Class_name"]
                    elif key in csv_object_name_dict:
                        new_name = csv_object_name_dict[key]["Class_name"]
                    else:
                        continue
                    data[new_name] = data.pop(key)

                with open(os.path.join(target_dir_path_sub, file_name), 'w') as f:
                    json.dump(data, f, indent=4)
                
                #### bbox side
                target_dir_path_sub = os.path.join(platform_path,"bbox","side_view_camera")
                with open(os.path.join(target_dir_path_sub, file_name), 'r') as f:
                    data = json.load(f)
                keys = list(data.keys())
                for key in keys:
                    if key in csv_old_name_dict:
                        new_name = csv_old_name_dict[key]["Class_name"]
                    elif key in csv_object_name_dict:
                        new_name = csv_object_name_dict[key]["Class_name"]
                    else:
                        continue
                    data[new_name] = data.pop(key)

                    with open(os.path.join(target_dir_path_sub, file_name), 'w') as f:
                        json.dump(data, f, indent=4)
                # print("bbox change complete : ", file_name)
                
                #### inst_seg top
                target_dir_path_sub = os.path.join(platform_path,"inst_seg","top_view_camera")
                with open(os.path.join(target_dir_path_sub, "semantics_mapping_"+file_name), 'r') as f:
                    data = json.load(f)

                keys = list(data.keys())
                for key in keys:
                    if data[key]["class"] in ["BACKGROUND","UNLABELLED" ]: continue
                    old_name = data[key]["class"]
                    if old_name in csv_old_name_dict:
                        new_name = csv_old_name_dict[old_name]["Class_name"]
                    elif old_name in csv_object_name_dict:
                        new_name = csv_object_name_dict[old_name]["Class_name"]
                    else:
                        continue
                    data[key]["class"] = new_name

                with open(os.path.join(target_dir_path_sub, "semantics_mapping_"+file_name), 'w') as f:
                    json.dump(data, f, indent=4)

                #### inst_seg side
                target_dir_path_sub = os.path.join(platform_path,"inst_seg","side_view_camera")
                with open(os.path.join(target_dir_path_sub, "semantics_mapping_"+file_name), 'r') as f:
                    data = json.load(f)
                keys = list(data.keys())
                for key in keys:
                    if data[key]["class"] in ["BACKGROUND","UNLABELLED" ]: continue
                    old_name = data[key]["class"]
                    if old_name in csv_old_name_dict:
                        new_name = csv_old_name_dict[old_name]["Class_name"]
                    elif old_name in csv_object_name_dict:
                        new_name = csv_object_name_dict[old_name]["Class_name"]
                    else:
                        continue
                    data[key]["class"] = new_name

                with open(os.path.join(target_dir_path_sub, "semantics_mapping_"+file_name), 'w') as f:
                    json.dump(data, f, indent=4)
                # print("inst_seg change complete : ", file_name)

                #### pregrasp
                target_dir_path_sub = os.path.join(platform_path,"pre_grasp")
                if os.path.exists(os.path.join(target_dir_path_sub, file_name)):
                    with open(os.path.join(target_dir_path_sub, file_name), 'r') as f:
                        data = json.load(f)

                    for idx,dd in enumerate(data):
                        for i,d in enumerate(dd["data"]):
                            old_name = d["target_object"]
                            if old_name in csv_old_name_dict:
                                new_name = csv_old_name_dict[old_name]["Class_name"]
                            elif old_name in csv_object_name_dict:
                                new_name = csv_object_name_dict[old_name]["Class_name"]
                            else:
                                continue
                            data[idx]["data"][i]["target_object"] = new_name
                                # print("changed : pregrasp", file_name)
                    with open(os.path.join(target_dir_path_sub, file_name), 'w') as f:
                        json.dump(data, f, indent=4)
                    
                
                #### grasp
                target_dir_path_sub = os.path.join(platform_path,"output_grasp")
                if os.path.exists(os.path.join(target_dir_path_sub, file_name)):
                    with open(os.path.join(target_dir_path_sub, file_name), 'r') as f:
                        data = json.load(f)

                    for idx,dd in enumerate(data):
                        old_name = dd["target_object"]
                        if old_name in csv_old_name_dict:
                            new_name = csv_old_name_dict[old_name]["Class_name"]
                        elif old_name in csv_object_name_dict:
                            new_name = csv_object_name_dict[old_name]["Class_name"]
                        else:
                            continue
                        data[idx]["target_object"] = new_name
                            # print("changed : grasp", file_name)
                    with open(os.path.join(target_dir_path_sub, file_name), 'w') as f:
                        json.dump(data, f, indent=4)



            

                    
                        

```
